# Customer Segmentation Using K-Means Clustering

Notebook ini dibuat sebagai portofolio Machine Learning untuk GitHub.

## 1. Import Library dan Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/customer_transaction.csv")
df.head()
```

## 2. Exploratory Data Analysis

```python
df.describe()
```

## 3. Feature Scaling dan Elbow Method

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

features = ["recency_days", "frequency", "monetary_value", "tenure_month", "visits_per_month"]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[features])

inertias = []
for k in range(2, 9):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertias.append(model.inertia_)

plt.plot(range(2, 9), inertias, marker="o")
plt.title("Elbow Method")
plt.xlabel("Jumlah Cluster")
plt.ylabel("Inertia")
plt.show()
```

## 4. K-Means Clustering

```python
from sklearn.decomposition import PCA

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X_scaled)

profile = df.groupby("cluster")[features].mean().round(2)
display(profile)

pca = PCA(n_components=2, random_state=42)
pca_result = pca.fit_transform(X_scaled)

plt.scatter(pca_result[:, 0], pca_result[:, 1], c=df["cluster"])
plt.title("Customer Segment Visualization")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.show()
```
